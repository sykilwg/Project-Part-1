```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import pandas as pd
import os
import time

# Setup
chromedriver_path = "C:/chromedriver.exe"
service = Service(executable_path=chromedriver_path)
driver = webdriver.Chrome(service=service)

movie_id = "tt1895587"  # Spotlight
movie_name = "Spotlight"
url = f"https://www.imdb.com/title/{movie_id}/reviews"
driver.get(url)

# Wait for reviews to load
time.sleep(3)

# Scroll to load more reviews
reviews = []
max_reviews = 100
scroll_pause_time = 2

while len(reviews) < max_reviews:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(scroll_pause_time)
    soup = BeautifulSoup(driver.page_source, "html.parser")
    review_blocks = soup.find_all("div", class_="ipc-html-content-inner-div")
    for block in review_blocks[len(reviews):max_reviews]:
        text = block.get_text(strip=True)
        if text:
            reviews.append(text)
    print(f"Collected {len(reviews)} reviews")

driver.quit()

# Save to folder
os.makedirs(movie_name, exist_ok=True)
df = pd.DataFrame(reviews, columns=["review"])
df.to_csv(f"{movie_name}_reviews.csv", index=False)
print(f"Saved to {movie_name}_reviews.csv")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
import pandas as pd
import time
import os

chromedriver_path = "C:/chromedriver.exe"
service = Service(executable_path=chromedriver_path)
driver = webdriver.Chrome(service=service)
actions = ActionChains(driver)

movie_id = "tt14230458"  
url = f"https://www.imdb.com/title/{movie_id}/reviews"
driver.get(url)

WebDriverWait(driver, 15).until(
    EC.presence_of_element_located((By.CLASS_NAME, "ipc-html-content-inner-div"))
)

reviews_collected = 0
max_reviews = 100
reviews = []
scroll_pause_time = 3  

folder_name = "Poor_Things_reviews"
os.makedirs(folder_name, exist_ok=True)

while reviews_collected < max_reviews:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(scroll_pause_time)  
    
    soup = BeautifulSoup(driver.page_source, "html.parser")
    
    review_blocks = soup.find_all("div", class_="ipc-html-content-inner-div")
    
    for block in review_blocks[reviews_collected:max_reviews]:
        try:
            review_text = block.get_text(strip=True)
            
            reviews.append(review_text)
            reviews_collected += 1
        except Exception as e:
            print("Error processing review:", e)
    
    print(f"Reviews collected: {reviews_collected}/{max_reviews}")

if reviews_collected > 0:
    df = pd.DataFrame(reviews, columns=["review"])
    df.to_csv(f"{folder_name}/Poor_Things_reviews.csv", index=False)
    print(f"Scraped {len(df)} reviews.")
else:
    print("No reviews were scraped.")

driver.quit()
```
